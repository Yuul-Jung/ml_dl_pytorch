# ch01_1. NumPy와 PyTorch

NumPy(넘파이, 수치 계산 라이브러리)와 PyTorch(파이토치, 딥러닝 프레임워크)의 기본 사용법을 같은 예제로 비교합니다.

## 개요

- NumPy의 `ndarray`(엔디에레이, 다차원 배열)와 PyTorch의 `Tensor`(텐서, 다차원 배열)를 구분합니다.
- PyTorch의 `Autograd`(오토그래드, 자동 미분)를 간단히 확인합니다.

## 1. NumPy와 PyTorch의 역할

| 구분 | NumPy | PyTorch |
|---|---|---|
| 핵심 자료형 | `ndarray` | `Tensor` |
| 주요 목적 | 수치 계산, 데이터 처리 | 딥러닝 모델 작성과 학습 |
| 자동 미분 | 지원하지 않음 | 지원 |
| GPU 연산 | 기본적으로 CPU 중심 | GPU 연산 지원 |
| 기본 문법 | 배열 중심 | NumPy와 매우 유사 |

```python
import numpy as np
import torch

print("NumPy version :", np.__version__)
print("PyTorch version:", torch.__version__)
```

## 2. 배열과 텐서 만들기

- `dtype`(디타입, 데이터 자료형)은 숫자를 저장하는 방식을 나타냅니다.
- 딥러닝에서는 일반적으로 `float32`(플로트 서티투, 32비트 실수형)를 많이 사용합니다.
- 두 자료형은 겉모습이 비슷하지만 서로 다른 라이브러리의 객체입니다.

```python
# 같은 데이터를 NumPy 배열과 PyTorch 텐서로 만듭니다.
numbers = [1, 2, 3, 4]

np_array = np.array(numbers, dtype=np.float32)
torch_tensor = torch.tensor(numbers, dtype=torch.float32)

print("NumPy 배열 :", np_array)
print("PyTorch 텐서:", torch_tensor)

print("NumPy 자료형 :", np_array.dtype)
print("PyTorch 자료형:", torch_tensor.dtype)
```

## 3. 기본 연산 비교

`+`, `-`, `*`, `/`는 기본적으로 **원소별 연산**입니다.

> **원소별 연산은 각 원소에 같은 계산을 적용합니다.**

```python
# 두 라이브러리의 기본 연산 문법은 거의 같습니다.
print("[NumPy]")
print("더하기 :", np_array + 10)
print("곱하기 :", np_array * 2)
print("합계   :", np_array.sum())
print("평균   :", np_array.mean())

print("\n[PyTorch]")
print("더하기 :", torch_tensor + 10)
print("곱하기 :", torch_tensor * 2)
print("합계   :", torch_tensor.sum())
print("평균   :", torch_tensor.mean())
```

## 4. Shape, 인덱싱, Reshape

- `shape`(셰이프, 데이터 모양)의 `(2, 3)`은 **2행 3열**을 의미합니다.
- 인덱스는 `0`부터 시작합니다.
- `reshape`는 데이터의 값과 개수는 유지하고 모양만 바꿉니다.
- 원소가 6개이므로 `(3, 2)`, `(1, 6)`, `(6, 1)` 등으로 변경할 수 있습니다.

```python
# 2행 3열 데이터
data = [[1, 2, 3],
        [4, 5, 6]]

np_matrix = np.array(data)
torch_matrix = torch.tensor(data)

print("NumPy shape :", np_matrix.shape)
print("PyTorch shape:", torch_matrix.shape)

# 첫 번째 행, 두 번째 열의 값
print("NumPy 값 :", np_matrix[0, 1])
print("PyTorch 값:", torch_matrix[0, 1])

# 2행 3열을 3행 2열로 변경
print("\nNumPy reshape")
print(np_matrix.reshape(3, 2))

print("\nPyTorch reshape")
print(torch_matrix.reshape(3, 2))
```

## 5. 원소별 곱셈과 행렬곱 구분

- `*` : Element-wise multiplication(원소별 곱셈)
- `@` : Matrix multiplication(행렬곱)

두 연산은 결과가 다르므로 반드시 구분해야 합니다.

```python
a_np = np.array([[1, 2],
                 [3, 4]])

b_np = np.array([[5, 6],
                 [7, 8]])

a_torch = torch.tensor(a_np)
b_torch = torch.tensor(b_np)

# 원소별 곱셈: 같은 위치의 원소끼리 곱합니다.
print("NumPy 원소별 곱셈")
print(a_np * b_np)

print("\nPyTorch 원소별 곱셈")
print(a_torch * b_torch)

# 행렬곱: 행과 열을 곱하여 새로운 행렬을 만듭니다.
print("\nNumPy 행렬곱")
print(a_np @ b_np)

print("\nPyTorch 행렬곱")
print(a_torch @ b_torch)
```

## 6. NumPy와 PyTorch 사이의 변환

- `torch.tensor(np_data)`는 NumPy 배열의 값을 이용해 새 텐서를 만듭니다.
- CPU 텐서는 `.numpy()`로 NumPy 배열로 변환할 수 있습니다.
- `.copy()`를 붙이면 원본 텐서와 독립된 NumPy 배열을 만들 수 있어 초보 실습에서 혼선을 줄일 수 있습니다.

```python
# NumPy 배열 → PyTorch 텐서
np_data = np.array([10, 20, 30], dtype=np.float32)
tensor_data = torch.tensor(np_data)

# PyTorch 텐서 → NumPy 배열
back_to_numpy = tensor_data.numpy().copy()

print("원본 NumPy :", np_data)
print("PyTorch 변환:", tensor_data)
print("다시 NumPy :", back_to_numpy)
```

## 7. PyTorch의 자동 미분

함수는 다음과 같습니다.

$$
y = x^2 + 2x
$$

미분 결과는 다음과 같습니다.

$$
\frac{dy}{dx} = 2x + 2
$$

`x = 3`을 대입하면 기울기는 `8`입니다. PyTorch는 `backward()`를 실행하여 이 값을 자동으로 계산합니다.

> **자동 미분은 신경망 학습에 필요한 기울기를 자동으로 계산합니다.**

```python
# x에 대한 계산 과정을 추적하도록 설정합니다.
x = torch.tensor(3.0, requires_grad=True)

# y = x² + 2x
y = x**2 + 2*x

# y를 x로 미분합니다. dy/dx
y.backward()

print("x =", x.item())
print("y =", y.item())
print("dy/dx =", x.grad.item())
```

## 8. 미니 실습: 센서 데이터 정규화

```python
# 센서 측정값을 0~1 범위로 변환합니다.
sensor_values = [20.0, 30.0, 40.0, 50.0]

# NumPy
np_sensor = np.array(sensor_values, dtype=np.float32)
np_normalized = (np_sensor - np_sensor.min()) / (np_sensor.max() - np_sensor.min())

# PyTorch
torch_sensor = torch.tensor(sensor_values, dtype=torch.float32)
torch_normalized = (
    (torch_sensor - torch_sensor.min())
    / (torch_sensor.max() - torch_sensor.min())
)

print("NumPy 결과 :", np_normalized)
print("PyTorch 결과:", torch_normalized)

# 두 결과가 같은지 확인합니다.
print("결과 일치:", np.allclose(np_normalized, torch_normalized.numpy()))
```

## 9. 핵심 정리

- NumPy의 기본 단위는 `ndarray`, PyTorch의 기본 단위는 `Tensor`입니다.
- `*`는 원소별 곱셈이고, `@`는 행렬곱입니다.
- PyTorch는 자동 미분을 지원하므로 딥러닝 모델 학습에 사용할 수 있습니다.
