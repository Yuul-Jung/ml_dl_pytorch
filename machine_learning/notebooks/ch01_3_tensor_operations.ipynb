# ch01_3. PyTorch Tensor 연산

## 개요

- 일정한 값과 난수로 Tensor(텐서, 다차원 배열)를 생성합니다.
- 여러 묶음의 센서 데이터를 하나로 결합합니다.

## 1. 라이브러리 불러오기

NumPy는 `np`, PyTorch(파이토치)는 `torch`라는 이름으로 사용합니다.

```python
import numpy as np
import torch
```

## 2. Tensor 생성 함수

PyTorch는 자주 사용하는 형태의 Tensor를 간단히 만들 수 있습니다.

| 함수 | 역할 |
|---|---|
| `torch.zeros()` | 모든 원소를 0으로 생성 |
| `torch.ones()` | 모든 원소를 1로 생성 |
| `torch.arange()` | 일정한 간격의 연속 값 생성 |
| `torch.rand()` | 0 이상 1 미만의 난수 생성 |
| `torch.randn()` | 평균 0, 표준편차 1인 난수 생성 |

`torch.manual_seed()`를 사용하면 다시 실행해도 같은 난수가 생성됩니다.

```python
# 같은 난수 결과를 얻기 위한 기준값
torch.manual_seed(42)

zeros_tensor = torch.zeros(2, 3)
ones_tensor = torch.ones(2, 3)
range_tensor = torch.arange(0, 6).reshape(2, 3)
random_tensor = torch.rand(2, 3)

print("zeros:")
print(zeros_tensor)

print()
print("ones:")
print(ones_tensor)

print()
print("arange + reshape:")
print(range_tensor)

print()
print("rand:")
print(random_tensor)
```

## 3. NumPy 배열과 Tensor의 메모리 공유

`torch.from_numpy()`로 만든 Tensor는 원본 NumPy 배열과 메모리를 공유할 수 있습니다.

따라서 NumPy 배열의 값을 바꾸면 Tensor의 값도 함께 바뀔 수 있습니다.

독립된 복사본이 필요할 때는 `clone()`을 사용합니다.

```python
numpy_data = np.array([[10, 20], [30, 40]], dtype=np.float32)

shared_tensor = torch.from_numpy(numpy_data)
copied_tensor = shared_tensor.clone()

# NumPy 배열의 첫 번째 값을 변경합니다.
numpy_data[0, 0] = 99

print("변경된 NumPy 배열:")
print(numpy_data)

print()
print("메모리를 공유하는 Tensor:")
print(shared_tensor)

print()
print("미리 복사한 독립 Tensor:")
print(copied_tensor)
```

## 4. CPU와 GPU 장치 선택

`device`(디바이스, 저장·연산 장치)는 Tensor가 저장되는 위치입니다.

- `cpu`: CPU 메모리
- `cuda`: NVIDIA(엔비디아) GPU 메모리

다음 코드는 GPU가 있으면 GPU를 선택하고, 없으면 CPU를 선택합니다.

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

data = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
data_on_device = data.to(device)

print("선택된 장치:", device)
print("Tensor의 장치:", data_on_device.device)
```

## 5. Tensor 모양 바꾸기

`reshape()`는 원소의 총개수를 유지하면서 Tensor의 모양을 바꿉니다.

$$
2 \times 3 = 3 \times 2 = 6
$$

따라서 원소가 6개인 Tensor는 `(2, 3)`, `(3, 2)`, `(1, 6)` 등의 모양으로 변경할 수 있습니다.

`flatten()`은 여러 차원의 Tensor를 1차원으로 펼칩니다.

```python
tensor_a = torch.arange(1, 7).reshape(2, 3)

print("원래 Tensor:")
print(tensor_a)
print("shape:", tensor_a.shape)

print()
print("reshape(3, 2):")
print(tensor_a.reshape(3, 2))

print()
print("flatten():")
print(tensor_a.flatten())
```

## 6. `torch.cat()`으로 연결하기

`torch.cat()`은 **기존 차원을 따라** Tensor를 연결합니다.

두 Tensor의 shape가 모두 `(2, 3)`일 때:

- `dim=0`: 첫 번째 차원의 크기가 증가 → `(4, 3)`
- `dim=1`: 두 번째 차원의 크기가 증가 → `(2, 6)`

> `dim`은 단순히 행·열 이름보다 **어느 차원의 크기를 늘리는가**로 이해하는 것이 좋습니다.

```python
tensor_b = torch.tensor(
    [[10, 20, 30],
     [40, 50, 60]]
)

cat_dim0 = torch.cat([tensor_a, tensor_b], dim=0)
cat_dim1 = torch.cat([tensor_a, tensor_b], dim=1)

print("cat dim=0:")
print(cat_dim0)
print("shape:", cat_dim0.shape)

print()
print("cat dim=1:")
print(cat_dim1)
print("shape:", cat_dim1.shape)
```

## 7. `torch.stack()`으로 쌓기

`torch.stack()`은 **새로운 차원을 추가하여** Tensor를 쌓습니다.

두 개의 `(2, 3)` Tensor를 `dim=0`으로 쌓으면 결과 shape는 `(2, 2, 3)`이 됩니다.

- 첫 번째 `2`: Tensor 묶음의 개수
- 두 번째 `2`: 각 Tensor의 행 개수
- 마지막 `3`: 각 Tensor의 열 개수

```python
stack_dim0 = torch.stack([tensor_a, tensor_b], dim=0)

print("stack dim=0:")
print(stack_dim0)
print("shape:", stack_dim0.shape)
```

## 8. `cat`과 `stack` 비교

| 함수 | 핵심 동작 | 예시 결과 |
|---|---|---|
| `torch.cat()` | 기존 차원을 따라 연결 | `(2, 3)` 두 개 → `(4, 3)` |
| `torch.stack()` | 새로운 차원을 만들어 쌓기 | `(2, 3)` 두 개 → `(2, 2, 3)` |

------------------------------------------------------------------------

> **`cat`은 Tensor를 연결하고, `stack`은 새로운 차원으로 묶습니다.**

## 9. 미니 실습: 센서 데이터 묶음 합치기

두 번에 나누어 측정한 온도·진동 데이터를 하나로 결합합니다.

각 행은 하나의 측정값이며, 각 열은 다음 의미를 가집니다.

- 첫 번째 열: 온도
- 두 번째 열: 진동

```python
# 첫 번째 측정 묶음
batch_1 = torch.tensor(
    [[22.0, 0.10],
     [25.0, 0.15]]
)

# 두 번째 측정 묶음
batch_2 = torch.tensor(
    [[31.0, 0.45],
     [36.0, 0.90]]
)

# 모든 측정값을 하나의 표로 연결합니다.
all_sensor_data = torch.cat([batch_1, batch_2], dim=0)

# 두 측정 묶음을 구분한 상태로 쌓습니다.
grouped_sensor_data = torch.stack([batch_1, batch_2], dim=0)

print("하나로 연결한 센서 데이터:")
print(all_sensor_data)
print("shape:", all_sensor_data.shape)

print()
print("측정 묶음별로 쌓은 센서 데이터:")
print(grouped_sensor_data)
print("shape:", grouped_sensor_data.shape)
```

## 11. 핵심 정리

- `zeros`, `ones`, `arange`, `rand`로 Tensor를 쉽게 생성할 수 있습니다.
- `.to(device)`는 Tensor를 CPU 또는 GPU로 이동합니다.
- `torch.stack()`은 새로운 차원을 추가하여 쌓습니다.
