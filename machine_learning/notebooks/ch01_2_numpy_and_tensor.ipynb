# ch01_2. NumPy 배열과 PyTorch Tensor 이해하기

## 개요

- Python(파이썬) 리스트와 NumPy(넘파이) 배열의 차이를 이해합니다.
- NumPy 배열을 PyTorch Tensor(파이토치 텐서, 다차원 배열)로 변환합니다.

## 1. NumPy와 PyTorch 불러오기

- NumPy는 `np`라는 짧은 이름으로 사용합니다.
- PyTorch는 `torch`라는 이름으로 사용합니다.

```python
import numpy as np
import torch
```

## 2. Python 리스트와 NumPy 배열 비교

Python 리스트에서 `+`는 두 리스트를 이어 붙입니다.
NumPy 배열에서 `+`는 같은 위치의 숫자를 더합니다.

> **NumPy 배열은 같은 위치의 원소끼리 수치 연산을 수행합니다.**

```python
# Python 리스트
list_a = [1, 2, 3]
list_b = [10, 20, 30]

# NumPy 배열
array_a = np.array([1, 2, 3])
array_b = np.array([10, 20, 30])

print("Python 리스트 + :", list_a + list_b)
print("NumPy 배열 +    :", array_a + array_b)
```

## 3. Scalar·Vector·Matrix 구분

| 구분 | 의미 | 예시 shape |
|---|---|---|
| Scalar(스칼라) | 값 1개 | `()` |
| Vector(벡터) | 1차원 데이터 | `(3,)` |
| Matrix(매트릭스) | 2차원 데이터 | `(2, 2)` |
| Tensor(텐서) | 3차원 이상의 데이터까지 포함 | `(2, 3, 4)` |

- `ndim`은 차원의 개수입니다.
- `shape`는 각 차원에 몇 개의 값이 있는지를 나타냅니다.

```python
scalar = np.array(7)
vector = np.array([1, 2, 3])
matrix = np.array([[1, 2], [3, 4]])

print("Scalar → ndim:", scalar.ndim, ", shape:", scalar.shape)
print("Vector → ndim:", vector.ndim, ", shape:", vector.shape)
print("Matrix → ndim:", matrix.ndim, ", shape:", matrix.shape)
```

## 4. 2차원 배열 읽기

다음 배열은 3행 2열 구조입니다.

$$
X =
\begin{bmatrix}
22.5 & 0.10 \\
31.0 & 0.45 \\
36.0 & 0.90
\end{bmatrix}
$$

각 행은 하나의 측정 데이터이고, 각 열은 측정 항목입니다.

- 첫 번째 열: 온도
- 두 번째 열: 진동

```python
sensor_data = np.array(
    [
        [22.5, 0.10],
        [31.0, 0.45],
        [36.0, 0.90],
    ],
    dtype=np.float32,
)

print(sensor_data)
print("차원 수:", sensor_data.ndim)
print("shape:", sensor_data.shape)
print("자료형:", sensor_data.dtype)
```

## 5. Indexing으로 원하는 값 선택하기

인덱스는 `0`부터 시작합니다.

- `sensor_data[0]` : 첫 번째 행
- `sensor_data[:, 0]` : 모든 행의 첫 번째 열
- `sensor_data[1, 0]` : 두 번째 행의 첫 번째 값

```python
print("첫 번째 측정 데이터:", sensor_data[0])
print("모든 온도값:", sensor_data[:, 0])
print("모든 진동값:", sensor_data[:, 1])
print("두 번째 온도값:", sensor_data[1, 0])
```

## 6. Boolean Mask로 조건에 맞는 값 선택하기

Boolean Mask(불리언 마스크)는 조건의 결과인 `True`와 `False`를 이용해 필요한 값만 선택합니다.

다음 예제에서는 온도가 30℃ 이상인 데이터만 선택합니다.

```python
temperatures = sensor_data[:, 0]

high_mask = temperatures >= 30.0
high_temperatures = temperatures[high_mask]

print("30℃ 이상 여부:", high_mask)
print("30℃ 이상 온도:", high_temperatures)
```

## 7. 간단한 Broadcasting 이해하기

Broadcasting(브로드캐스팅, 작은 데이터를 큰 배열에 맞춰 자동 확장하는 기능)을 사용하면 모든 값에 같은 숫자를 더하거나 곱할 수 있습니다.

센서 보정값 `1.5`를 모든 온도에 더해 보겠습니다.

```python
temperature_offset = 1.5
corrected_temperatures = temperatures + temperature_offset

print("보정 전:", temperatures)
print("보정 후:", corrected_temperatures)
```

## 8. NumPy 배열을 PyTorch Tensor로 변환하기

`torch.from_numpy()`를 사용하면 NumPy 배열을 Tensor로 변환할 수 있습니다.

두 자료형은 모양이 비슷하지만 소속 라이브러리가 다릅니다.

- NumPy: `ndarray`
- PyTorch: `Tensor`

```python
sensor_tensor = torch.from_numpy(sensor_data)

print(sensor_tensor)
print("자료형:", type(sensor_tensor))
print("dim:", sensor_tensor.dim())
print("shape:", sensor_tensor.shape)
print("dtype:", sensor_tensor.dtype)
```

## 9. NumPy와 Tensor의 속성 비교

| 확인 항목 | NumPy 배열 | PyTorch Tensor |
|---|---|---|
| 차원 수 | `ndim` | `dim()` |
| 모양 | `shape` | `shape` |
| 자료형 | `dtype` | `dtype` |
| 저장 장치 | 별도 확인 없음 | `device` |

PyTorch Tensor는 CPU(씨피유) 또는 GPU(지피유)에 저장할 수 있으므로 `device` 속성이 있습니다.

```python
print("NumPy ndim :", sensor_data.ndim)
print("Tensor dim :", sensor_tensor.dim())

print("NumPy shape:", sensor_data.shape)
print("Tensor shape:", sensor_tensor.shape)

print("Tensor device:", sensor_tensor.device)
```

## 10. 미니 실습: 센서 데이터 판정하기

온도와 진동 데이터를 Tensor로 변환한 뒤 다음 조건을 확인합니다.

- 온도 35℃ 이상
- 진동 0.7 이상

두 조건이 모두 참이면 점검이 필요한 데이터로 판정합니다.

```python
practice_data = np.array(
    [
        [24.0, 0.20],
        [34.0, 0.55],
        [38.0, 0.85],
    ],
    dtype=np.float32,
)

practice_tensor = torch.from_numpy(practice_data)

temperature_alarm = practice_tensor[:, 0] >= 35.0
vibration_alarm = practice_tensor[:, 1] >= 0.7
inspection_required = temperature_alarm & vibration_alarm

print("센서 데이터:")
print(practice_tensor)

print()
print("점검 필요 여부:")
print(inspection_required)

print()
print("점검이 필요한 데이터:")
print(practice_tensor[inspection_required])
```

## 12. 핵심 정리

- Python 리스트는 일반적인 데이터 저장에 사용합니다.
- 불리언 마스크는 조건에 맞는 데이터를 선택합니다.
- PyTorch Tensor는 `dim`, `shape`, `dtype`, `device` 속성을 확인할 수 있습니다.
